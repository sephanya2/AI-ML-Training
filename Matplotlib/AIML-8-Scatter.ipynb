```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(r"C:\Users\Sephanya\Downloads\googleplaystore.csv",nrows = 1000)

df
```

```python
x = df['Rating']
y = df['Reviews']

plt.scatter(x,y,c='g',marker='*',edgecolor='green',linewidth = 2)
plt.xlabel('Rating')
plt.ylabel('Review')
plt.title('Google play store dataset')
plt.figure(figsize = (16,9))
plt.show()
```

```python
x = df['Rating']
y = df['Reviews']

plt.scatter(x,y,c='teal',marker='*',edgecolor='green',linewidth = 2,s=80,alpha=0.5)
plt.scatter(x, df['Installs'],c='teal',marker='*', edgecolor='pink',linewidth = 2,alpha=0.5)
plt.xlabel('Rating')
plt.ylabel('Review')
plt.title('Google play store dataset')
plt.figure(figsize = (16,9))
plt.show()
```
