```python
# pie chart
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(r"C:\Users\Sephanya\Downloads\googleplaystore.csv", nrows=1000)

category = df["Category"].value_counts().head(10)

colors = [
    'green','teal','yellow','salmon','lightsalmon',
    'skyblue','orange','pink','violet','gold'
]

explode = [0]*len(category)
explode[0] = 0.1

plt.figure(figsize=(8,8))

plt.pie(
    category,
    labels=category.index,
    autopct='%1.1f%%',
    startangle=90,
    explode=explode,
    colors=colors,
    shadow=True,
    textprops={'fontsize':10},
    pctdistance=0.7,
    labeldistance=1.1
)

plt.title("Top 10 App Categories")
plt.show()
```
