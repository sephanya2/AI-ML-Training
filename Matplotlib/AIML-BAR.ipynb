```python
import matplotlib.pyplot as plt
import numpy as np

classes = ['Python', 'R', 'AI', 'ML', 'DS']
classes_1 = [90, 60, 70, 39, 40]
classes_2 = [79, 4, 29, 50, 80]
classes_3 = [9, 4, 29, 30, 60]

# X-axis positions
x = np.arange(len(classes))

# Width of each bar
width = 0.25
plt.figure(figsize=(8,5))

plt.bar(classes_1,  color='green', edgecolor='black',
        label='Class 1 Students')

plt.bar(,color='teal', edgecolor='black',
        label='Class 2 Students')

# Labels and title
plt.title("Students in Classes", fontsize=15)
plt.xlabel("Classes", fontsize=12)
plt.ylabel("Number of Students", fontsize=12)

# Display class names instead of numbers
plt.xticks(x, classes,rotation = 90)
#plt.legend()
#plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

classes = ['Python', 'R', 'AI', 'ML', 'DS']
classes_1 = [90, 60, 70, 39, 40]
classes_2 = [79, 4, 29, 50, 80]
classes_3 = [9, 4, 29, 30, 60]

x = np.arange(len(classes))
width = 0.25

plt.figure(figsize=(8,5))

plt.bar(x, classes_1, width=width, color='green', edgecolor='black', label='Class 1')
plt.bar(x + width, classes_2, width=width, color='teal', edgecolor='black', label='Class 2')
plt.bar(x + 2*width, classes_3, width=width, color='yellow', edgecolor='black', label='Class 3')

# Move tick labels to the center
plt.xticks(x + width, classes)

plt.title("Students in Classes")
plt.xlabel("Classes")
plt.ylabel("Number of Students")
#plt.legend()

plt.show()
```
