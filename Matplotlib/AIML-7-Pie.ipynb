```python
import matplotlib.pyplot as plt

classes = ['Python', 'R', 'AI', 'ML', 'DS']
classes_1 = [90, 60, 70, 39, 40]
colors = ['green','teal','yellow','Salmon','LightSalmon']
wedgeprops={'linewidth':4,'width':1,'edgecolor':'k'}

plt.figure(figsize=(7,7))

plt.pie(
    classes_1,
    labels=classes,
    autopct='%1.1f%%',      # Show percentage
    startangle=90,          # Start from top
    explode=(0.1, 0, 0, 0, 0),  # Highlight Python
    colors = colors,
    radius = 1,
    shadow=True,
    textprops = {'fontsize':20},
    pctdistance = 0.6,
    labeldistance = 1.4,
    counterclock = True,
    wedgeprops = wedgeprops,
    center = (2,3),
    frame = True,
    rotatelabels = True
)

plt.title("Class 1 Students Distribution")
plt.legend(title="Subjects", loc="upper right")

plt.show()
```
